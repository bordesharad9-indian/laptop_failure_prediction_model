```python

# 3. REMOVE DUPLICATE ROWS

data = data.drop_duplicates()

print("\nShape after removing duplicates:", data.shape)

```

```python
import pandas as pd
# 4. CONVERT TIMESTAMP TO DATETIME
data["Timestamp"] = pd.to_datetime(data["Timestamp"], errors="coerce")

# Check invalid timestamps
print("\nInvalid timestamps:",
      data["Timestamp"].isnull().sum())

```

```python






# 7. CHECK DATA TYPES


numeric_columns = [
    "Timestamp",
    "CPU_Temp_C",
    "GPU_Temp_C",
    "CPU_Usage_Pct",
    "GPU_Usage_Pct",
    "RAM_Usage_Pct",
    "Disk_Usage_Pct",
    "Disk_IO_MBps",
    "Battery_Level_Pct",
    "Fan_Speed_RPM",
    
]

target_column = "Failure_Status"

# Convert numeric columns to numeric values
for col in numeric_columns + [target_column]:
    data[col] = pd.to_numeric(data[col], errors="coerce")

```

```python
from sklearn.impute import SimpleImputer

# 8. HANDLE MISSING VALUES
# Median imputation for numerical features
imputer = SimpleImputer(strategy="median")

data[numeric_columns] = imputer.fit_transform(
    data[numeric_columns]
)

# Remove rows where target is missing
df = data.dropna(subset=[target_column])

```

```python
df = pd.get_dummies(df, columns=["System_Event"], dtype=int)
```

```python
df.to_csv("system_telemetry_dataset.csv", index=False)
```
